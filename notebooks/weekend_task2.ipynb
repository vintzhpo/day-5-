```python
import pandas as pd 
import numpy as np 
from sklearn.model_selection import train_test_split
df=pd.read_csv('../data/features_df.csv')

train, test = train_test_split(df, test_size=0.2, stratify=df["is_churned"], random_state=42)

print('--Churn % Харьцуулалт--')
print(f"Нийт датан дахь цуцлалт:{df['is_churned'].mean()*100:.2f}%")
print(f"Сургалтын датан дахь цуцлалт:{train['is_churned'].mean()*100:.2f}%")
print(f"Туршилтын датан дахь цуцлалт:{test['is_churned'].mean()*100:.2f}%")

dumb_baseline_accuracy = (test['is_churned'] ==0).mean()*100
print (f'Dumb Baseline Accuracy: {dumb_baseline_accuracy:.2f}%')
```
